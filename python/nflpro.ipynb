```python
for g in game_id_elias[[6]]:
        print(g)

        pbp_data = nflpro_api.get_play_by_play_stats(
                game_id=g
        )
        if pbp_data:
                print("Play-by-Play:")
                print(pbp_data)

        pbp_df = nflpro_api.stats_to_dataframe(pbp_data)

pbp_df
```

```python
import pandas as pd

data_list = pbp_data["plays"]

important_cols = []
unnest_cols = ["playStats", "unofficialPlay", "nflIds", "offense", "defense", "passInfo", "recInfo"]

if not data_list:
    pd.DataFrame()

formatted_df = pd.DataFrame(data_list)

for col in unnest_cols:
        if col in formatted_df.columns:
            # Handle lists by extracting the first element if it exists
            formatted_df[col] = formatted_df[col.apply(
                lambda x: x[0] if isinstance(x, list) and len(x) > 0 else x
            )]

            # Normalize only if the column contains dictionaries after list handling
            if formatted_df[col].apply(lambda x: isinstance(x, dict)).any():
                formatted_df = formatted_df.join(
                    pd.json_normalize(formatted_df[col]).add_prefix(f"{col}_"),
                    rsuffix=f"_{col}",
                ).drop(col, axis=1)
            else:
                print(f"Skipping json_normalize for {col} as it does not contain dictionaries.")

remaining_cols = [
    col for col in formatted_df.columns if col not in important_cols
]
ordered_cols = [
    col for col in important_cols if col in formatted_df.columns
] + remaining_cols

formatted_df[ordered_cols]
```

```python
plays_list = pbp_data.get(
    "plays",
    [],
)  # Access the list of play dictionaries

plays_data = []  # Initialize an empty list to hold the data

for play in plays_list:
    row = {
        "gameId": play["gameId"],
        "playId": play["playId"],
        "playDescription": play["playDescription"],
        "possessionTeam": play["possessionTeam"],
        "down": play["down"],
        "yardsToGo": play["yardsToGo"],
        "quarter": play["quarter"],
    }

    plays_data.append(row)

plays_data
```
